```python
import numpy as np
import matplotlib.pyplot as plt

class Percolation:
    def __init__(self, n):
        self.n = n
        self.grid = np.zeros((n, n), dtype=bool)
        self.full = np.zeros((n, n), dtype=bool)

    def open_site(self, row, col):
        self.grid[row, col] = True

    def is_open(self, row, col):
        return self.grid[row, col]

    def is_full(self, row, col):
        return self.full[row, col]

    def percolates(self):
        for col in range(self.n):
            if self.full[self.n - 1, col]:
                return True
        return False

    def flow(self):
        for col in range(self.n):
            self._flow(0, col)

    def _flow(self, row, col):
        if row < 0 or row >= self.n or col < 0 or col >= self.n:
            return
        if not self.is_open(row, col) or self.is_full(row, col):
            return
        self.full[row, col] = True
        self._flow(row + 1, col)
        self._flow(row - 1, col)
        self._flow(row, col + 1)
        self._flow(row, col - 1)

def simulate_percolation(n, p):
    perc = Percolation(n)
    for row in range(n):
        for col in range(n):
            if np.random.rand() < p:
                perc.open_site(row, col)
    perc.flow()
    return perc

def plot_percolation(perc):
    plt.imshow(perc.grid, cmap='Greys', interpolation='none')
    plt.imshow(perc.full, cmap='Blues', alpha=0.5, interpolation='none')
    plt.title('Percolation Simulation')
    plt.show()

# Parameters
n = 20  # Grid size
p = 0.6  # Probability of site being open

# Simulate and plot percolation
perc = simulate_percolation(n, p)
plot_percolation(perc)
```
